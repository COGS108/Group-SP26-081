# Research Question

In 2010, top-selling AAA games like Mass Effect 2, Call of Duty: Black Ops, and Starcraft 2 ran on hardware like Intel Core 2 Duos with 1.8 GHZ, over 2 GB of RAM, and 256 MB/NVIDIA GeForce 8600 GPU. How does the historical shift of listed minimum CPU, GPU, and RAM requirements for such top-selling AAA games compare to the median user specifications of Steam users from 2010 to 2020? Furthermore, is the gap between minimum required specifications and median consumer hardware widening over time, indicating a growing barrier to entry for the average PC gamer?

## Background and Prior Work


From 2010 to 2020, the PC gaming landscape underwent a major transformation as AAA games increased in scale, graphical fidelity, and simulation complexity. These advances led developers to steadily raise minimum system requirements, requiring more powerful CPUs, GPUs, and larger amounts of RAM. However, this rapid increase in software demands did not perfectly align with consumer hardware adoption. While developers pushed toward higher performance baselines, many PC gamers continued using mid-range systems for extended periods, creating a growing gap between required and available hardware.

A key factor behind rising requirements was the transition toward multi-core CPU utilization in modern game engines. Earlier titles primarily relied on single-core performance, but by the early 2010s, developers began optimizing engines to take advantage of multiple cores for tasks such as physics simulation, artificial intelligence, and world streaming. This shift increased the baseline expectation for CPUs, as games increasingly depended on parallel processing rather than raw clock speed. At the same time, improvements in console hardware also influenced PC requirements, as cross-platform development pushed developers to target higher-performance systems overall<a href="https://www.club386.com/game-system-requirements-evolution/">[1]</a>.

In parallel, GPU requirements increased significantly due to advancements in rendering techniques. Higher-resolution textures, dynamic lighting, and more complex shader systems placed greater strain on graphics hardware. Historical analyses of GPU performance and gaming requirements show a clear upward trend in memory and processing demands, with modern AAA titles often requiring several gigabytes of VRAM and significantly more processing power than earlier games<a href="https://www.techspot.com/article/2902-nvidia-gpu-configuration-history/">[2]</a>. These graphical improvements were a major driver of hardware upgrades throughout the latter half of the decade.

Despite these increasing requirements, data from the Steam Hardware Survey demonstrates that consumer hardware adoption progressed more gradually. The survey, which collects hardware data from millions of users, is widely used by developers to understand the distribution of real-world PC configurations<a href="https://store.steampowered.com/hwsurvey/Steam-Hardware-Software-Survey-Welcome-to-Steam">[3]</a>. Even in recent years, mid-range GPUs and moderate system configurations have remained the most common among players, rather than the high-end hardware typically recommended for modern AAA titles. This highlights a persistent mismatch between developer expectations and the hardware that most players actually own, reinforcing the accessibility challenges created by rising system requirements.

This dataset contains information on the minimum and recommended hardware requirements for a large set of PC video games. It includes metrics describing the CPU, GPU, system RAM, VRAM, and operating system, all of which represent the baseline hardware performance needed to run each game. CPU clock speed and turbo speed are measured in gigahertz (GHz), indicating how many billions of cycles per second the processor can execute. Higher clock speeds generally correspond to faster processing and better performance. System RAM and GPU VRAM are reported in megabytes (MB) or gigabytes (GB). During data cleaning, these values were standardized to MB to ensure consistency. RAM determines how much data the system can hold in active memory, while VRAM specifically stores textures, frame buffers, and other graphical assets.

This data also included GPU metrics. Shader model and DirectX version indicate the graphical feature set the GPU supports, which determines if a game can run certain lighting models, shading, or rendering pipelines. GPU boost clock, measured in megahertz (MHz), reflects the peak operating frequency the GPU can reach under load. GPU memory bandwidth, measured in gigabytes per second (GB/s), describes how quickly the GPU can transfer data to the VRAM and processing cores.

A major limitation of this dataset is that all entries were scraped from a single public website, rather than from multiple independent sources. Relying on one source can have strong source bias, because the completeness, accuracy, and formatting of data depend on how the site presents the requirements.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
def clean_memory(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().replace(" ", "")
    if "gb" in x:
        return float(x.replace("gb", "")) * 1024
    if "mb" in x:
        return float(x.replace("mb", ""))
    return np.nan

def numeric_clean(x):
    if pd.isna(x):
        return np.nan
    x = str(x)
    for unit in ["ghz", "mhz", "hz", ","]:
        x = x.replace(unit, "")
    return pd.to_numeric(x, errors = "coerce")

## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
#Step 1: Loading in the dataset
df1 = pd.read_csv('data/00-raw/videogame_requirements[1].csv')

#Step 2: Check the structure of the dataset
df1.info()

#Step 3: Get the size of the dataset
print("Original dataset shape:\n", df1.shape)

#Step 4: finding how many missing values
total=df1.isna().sum()

#figuring out the percentage
precentage=(df1.isna().mean()*100).round(2)

#how many rows are missing values
missing_rows = df1[df1.isna().any(axis=1)]

print("Total values missing:\n", total)
print("Shape of all the missing values:\n", missing_rows.shape)

#finding if the missing value is random or not
missing_indicators = df1.isna().astype(int)
missing_indicators.corr()
sns.heatmap(missing_indicators.corr(), annot = False, cmap = 'viridis')
plt.show()

#converting object data to numbers
df1['Release_Date'] = pd.to_datetime(df1['Release_Date'], errors = 'coerce')
df1['Release_Year'] = df1['Release_Date'].dt.year

df1['Min_RAM_MB'] = df1['Min_RAM'].apply(clean_memory)
df1['Min_VRAM_MB'] = df1['Min_VRAM'].apply(clean_memory)

df1['Min_GPU_Memory'] = df1['Min_GPU_Memory'].apply(numeric_clean)
df1['Min_GPU_Boost_Clock'] = df1['Min_GPU_Boost_Clock'].apply(numeric_clean)
df1['Min_GPU_Memory_Bandwidth'] = df1['Min_GPU_Memory_Bandwidth'].apply(numeric_clean)

#Step 5: finding any outliers
numeric_cols = df1.select_dtypes(include=[float, int]).columns

# finding the z-score
z = np.abs((df1[numeric_cols] - df1[numeric_cols].mean()) / df1[numeric_cols].std())
outlier = df1[(z>3).any(axis=1)]
print("Shape of all the outliers:\n", outlier.shape)

#flagging the oulier by adding a column
df1['outlier_flag'] = (z>3).any(axis=1)

#keeping columns that are relevant
kept_cols = ["Name", "Release_Date", "Release_Year", "Min_CPU_CPU_Speed",
             "Min_CPU_Turbo_Speed", "Min_CPU_Physical_Cores", "Min_CPU_Threads",
             "Min_CPU_GD_Rating", "Min_GPU_Shader", "Min_GPU_Direct_X", "Min_GPU_Memory",
             "Min_GPU_GD_RATING", "Min_GPU_Boost_Clock", "Min_GPU_Memory_Bandwidth", "Min_GPU_Process",
             "Min_RAM_MB", "Min_VRAM_MB"]
df1_reduced = df1[kept_cols].copy()

#Step 6: cleaning up the data
clean_df1 = df1_reduced.dropna(how='any').copy()
clean_df1.to_csv('data/01-interim/videogame_requirements_cleaned.csv', index=False)

#processing the final dataset after cleaning it
processed_df1= clean_df1.copy()
processed_df1.to_csv('data/02-processed/videogame_requirements_processed.csv', index=False)

print(processed_df1.head())
```

This dataset contains monthly hardware survey information collected from Steam users between 2010 and 2020. The Steam Hardware Survey is conducted by Valve and records information about the computer hardware used by players on the Steam platform. For this project, the dataset was filtered to focus specifically on hardware categories relevant to gaming performance: CPU speed, system RAM, and video card descriptions. These variables are important because the project aims to compare the minimum hardware requirements of AAA PC games with the hardware commonly owned by average Steam users over time.

The dataset has been cleaned and tidied. The date column was converted into a datetime format and a separate year column was created to simplify yearly comparisons. The dataset was then filtered to only include observations from 2010 through 2020 to match the scope of the research question. Categories unrelated to gaming hardware were removed, leaving only CPU, RAM, and GPU information. Missing values were inspected and rows missing hardware names were removed because they could not be meaningfully analyzed. Entries with invalid percentage values outside the range of 0 to 1 were also removed. In addition, generic GPU labels such as “Other” were excluded because they were too broad to provide useful information about actual consumer hardware trends.


The biggest concern that comes with ths dataset is that the Steam Hardware Survey only represents Steam users and not the entire population of PC gamers, this could introduce some sampling bias because Steam users may differ from other PC gamers. Another issue that this dataset represents is that GPU performance is difficult to measure directly from categorical data of model names alone, comparing GPUs numerically is more complicated than say comparing CPU speeds that have a quantifiable measure.

```python
import pandas as pd

Steam_HW_Survey = pd.read_csv('data/00-raw/shs.csv')

#print out original shape to compare to clean/tidy shape
print("Original dataset shape:")
print(Steam_HW_Survey.shape)

#convert date column to datetime and under our time constraint of 2010-2020
Steam_HW_Survey['date'] = pd.to_datetime(Steam_HW_Survey['date'])

Steam_HW_Survey = Steam_HW_Survey[
    (Steam_HW_Survey['date'].dt.year >= 2010) &
    (Steam_HW_Survey['date'].dt.year <= 2020)
]

#ensure relevant categories exist in dataset
categories = [
    'AMD CPU Speeds',
    'Intel CPU Speeds',
    'System RAM',
    'Video Card Description'
]

Steam_HW_Survey = Steam_HW_Survey[Steam_HW_Survey['category'].isin(categories)]

Steam_HW_Survey['year'] = Steam_HW_Survey['date'].dt.year

print("\nRelevant categories:")
print(Steam_HW_Survey['category'].unique())

print("\nMissing values:")
print(Steam_HW_Survey.isnull().sum())

print("\nRows with missing values:")
print(Steam_HW_Survey[Steam_HW_Survey.isnull().any(axis=1)].shape[0])

print("\nSuspicious percentage values:")
print(Steam_HW_Survey[(Steam_HW_Survey['percentage'] < 0) | (Steam_HW_Survey['percentage'] > 1)].shape[0])

Steam_HW_Survey = Steam_HW_Survey[(Steam_HW_Survey['percentage'] >= 0) & (Steam_HW_Survey['percentage'] <= 1)]

Steam_HW_Survey = Steam_HW_Survey.dropna(subset=['name'])

Steam_HW_Survey['category'] = Steam_HW_Survey['category'].str.strip()
Steam_HW_Survey['name'] = Steam_HW_Survey['name'].str.strip()

Steam_HW_Survey = Steam_HW_Survey[Steam_HW_Survey['name'] != 'Other']

#display clean/tidy shape for comparison to original set
print("\nCleaned dataset shape:")
print(Steam_HW_Survey.shape)

print("\nCleaned dataset preview:")
display(Steam_HW_Survey.head())

#display hardware grouped by our relevant categories for each year by their popularity
print("\nMost common hardware by category and year:")

top_hardware = Steam_HW_Survey.loc[
    Steam_HW_Survey.groupby(['year', 'category'])['percentage'].idxmax()
]

top_hardware = top_hardware[
    ['year', 'category', 'name', 'percentage']
].sort_values(['year', 'category'])

display(top_hardware.head(20))

Steam_HW_Survey.to_csv('data/02-processed/cleaned_shs.csv', index=False)
```

#### Steam User Hardware Trends

This section examines the hardware commonly owned by Steam users between 2010 and 2020 using the Steam Hardware Survey dataset. The dataset records the percentage of users with different CPU speeds and RAM capacities

To compare this dataset with the game requirements dataset, CPU speed ranges were converted into numerical midpoint values. Weighted averages were then calculated using the percentage of Steam users represented by each hardware category.

The purpose of this analysis is to determine how consumer hardware evolved during the same period and whether it kept pace with rising game requirements.

```python
# Since the CPU speeds are given as ranges per year we need to convert them to numbers
# To do this we simply added the two numbers and divided by two
import re

def extract_midpoint(cpu_range):
    nums = re.findall(r"\d+\.\d+", str(cpu_range))

    if len(nums) >= 2:
        return (float(nums[0]) + float(nums[1])) / 2

    return None

# Next we find the weighted average CPU speed by year
steam_cpu = Steam_HW_Survey[
    Steam_HW_Survey["category"].isin([
        "AMD CPU Speeds",
        "Intel CPU Speeds"
    ])
].copy()

steam_cpu["cpu_midpoint"] = (
    steam_cpu["name"]
    .apply(extract_midpoint)
)

steam_cpu_yearly = (
    steam_cpu.groupby("year")
    .apply(lambda x:
        (x["cpu_midpoint"] * x["percentage"]).sum()
        / x["percentage"].sum()
    )
)

# We then plot these hardware trends
steam_cpu_yearly.plot(
    figsize=(10,5),
    title="Average Steam User CPU Speed by Year"
)
```

### Hardware Gap Between AAA Games and Steam Users
In this section, we directly compare the minimum hardware requirements of AAA games with the hardware commonly owned by Steam users between 2010 and 2020.

To perform this comparison, yearly median game hardware requirements were merged with yearly weighted average Steam hardware statistics. This allows us to measure the gap between the hardware required to run modern games and the hardware most users actually owned during the same years.

If the gap increases over time, this would support the hypothesis that AAA game requirements grew faster than consumer hardware adoption.

```python
# First we create yearly CPU game requirements 
game_cpu_yearly = (
    processed_df1.groupby("Release_Year")["Min_CPU_CPU_Speed"]
    .median()
    .reset_index(name="median_game_cpu")
)

game_cpu_yearly.head()
```

```python
# We can then merge this with the Steam hardware data
steam_cpu_yearly = (
    steam_cpu.groupby("year")
    .apply(lambda x:
        (x["cpu_midpoint"] * x["percentage"]).sum()
        / x["percentage"].sum()
    )
    .reset_index(name="avg_steam_cpu")
)

comparison = pd.merge(
    game_cpu_yearly,
    steam_cpu_yearly,
    left_on="Release_Year",
    right_on="year"
)

comparison.head()
```

```python
# We can then compute the gap between the median and average of pc users

comparison["cpu_gap"] = (
    comparison["median_game_cpu"]
    - comparison["avg_steam_cpu"]
)

comparison.head()
```

```python

plt.figure(figsize=(12,6))

plt.plot(
    comparison["Release_Year"],
    comparison["median_game_cpu"],
    marker="o",
    label="Median Game CPU Requirement"
)

plt.plot(
    comparison["Release_Year"],
    comparison["avg_steam_cpu"],
    marker="o",
    label="Average Steam User CPU Speed"
)

plt.title("AAA Game CPU Requirements vs Steam User CPU Speeds")
plt.xlabel("Year")
plt.ylabel("CPU Speed (GHz)")

plt.legend()
plt.grid(True)

plt.show()
```

The graph shows that from 2010 to around 2015, the average CPU speed owned by Steam users was generally higher than the median minimum CPU speed required by AAA games. During this period, most users appeared to have hardware capable of meeting or exceeding typical game requirements.

However around 2016 the median minimum CPU requirements for AAA games increased rapidly, while the average Steam user CPU speed remained relatively stable. By 2017 and onward, game CPU requirements consistently exceeded the average CPU speeds of Steam users.

This widening gap suggests that AAA game hardware requirements were increasing faster than consumer hardware adoption during the later years of the study. These finds support our original hypothesis that the minimum requirements of games have increased at a faster rate than the average Steam user.
