## Abstract

This project examined whether a gap has emerged between the median consumer hardware and the minimum system requirements for Steam games from 2010 to 2020. Understanding this gap is important because it reflects the real-world accessibility of PC gaming. Many people continue using older or budget hardware due to cost, availability, and upgrade habits, while modern games increasingly target newer architectures and higher performance baselines.

To evaluate this, we cleaned and standardized the two datasets we had so that we could compare CPU speed, GPU performance, and RAM directly. We first visualized each dataset independently using line plots to observe trends over the year in the median. Box plots were then used to examine distribution range and identify outliers that might distort trend interpretation. After establishing the internal structure of each dataset, we overlaid both sets of values on a combined line plot to be able to directly compare game requirements growth against consumer hardware growth over the decade.

The results show that game requirements increased more rapidly than the hardware owned by the median Steam user. CPU and RAM requirements rose steadily, while consumer hardware improved more slowly and inconsistently. This produced a widening performance gap over time, indicating that modern games have become progressively less accessible to players with mid-range or aging systems. Overall, the analysis supports the conclusion that the hardware expectations of game developers have outpaced the upgrade trends of typical PC users.

## Background and Prior Work


From 2010 to 2020, the PC gaming landscape underwent a major transformation as AAA games increased in scale, graphical fidelity, and simulation complexity. These advances led developers to steadily raise minimum system requirements, requiring more powerful CPUs, GPUs, and larger amounts of RAM. However, this rapid increase in software demands did not perfectly align with consumer hardware adoption. While developers pushed toward higher performance baselines, many PC gamers continued using mid-range systems for extended periods, creating a growing gap between required and available hardware. On the other side, recent development and cultural appreciation of indie gaming has established a modern paradigm that contrasts the issues AAA games present. Independent gaming developers have increasingly proven to prioritize artistry, uniqueness, and overall accessibility compared to the widening gap that has further emerged from the demands of modern AAA games.

A key factor behind rising requirements was the transition toward multi-core CPU utilization in modern game engines. Earlier titles primarily relied on single-core performance, but by the early 2010s, developers began optimizing engines to take advantage of multiple cores for tasks such as physics simulation, artificial intelligence, and world streaming. This shift increased the baseline expectation for CPUs, as games increasingly depended on parallel processing rather than raw clock speed. At the same time, improvements in console hardware also influenced PC requirements, as cross-platform development pushed developers to target higher-performance systems overall<a href="https://www.club386.com/game-system-requirements-evolution/">[1]</a>. Indie game development often relied and continues to rely on general-purpose engines that are widely accessible by Steam users. Not only does this frequently avoids budget issues, it ensures that people have access to their games.

In parallel, GPU requirements increased significantly due to advancements in rendering techniques. Higher-resolution textures, dynamic lighting, and more complex shader systems placed greater strain on graphics hardware. Historical analyses of GPU performance and gaming requirements show a clear upward trend in memory and processing demands, with modern AAA titles often requiring several gigabytes of VRAM and significantly more processing power than earlier games<a href="https://www.techspot.com/article/2902-nvidia-gpu-configuration-history/">[2]</a>. These graphical improvements were a major driver of hardware upgrades throughout the latter half of the decade.

Conversely, the market for indie games use of commercial engines like Unity or Unreal Engine allowed them to thrive on different graphical standards. Stylization of uncommon visual styles like 2D and pixel art gives them creative freedom plus focus on specialization of other features of their games. They essentially bypass the need for heavier VRAM hardware that can easily become costly.

Despite these increasing requirements, data from the Steam Hardware Survey demonstrates that consumer hardware adoption progressed more gradually. The survey, which collects hardware data from millions of users, is widely used by developers to understand the distribution of real-world PC configurations<a href="https://store.steampowered.com/hwsurvey/Steam-Hardware-Software-Survey-Welcome-to-Steam">[3]</a>. Even in recent years, mid-range GPUs and moderate system configurations have remained the most common among players, rather than the high-end hardware typically recommended for modern AAA titles. This highlights a persistent mismatch between developer expectations and the hardware that most players actually own, reinforcing the accessibility challenges created by rising system requirements. And despite this contrast between players and game demands, indie games balance out the disorder by usually fitting within the scope of Steam user's hardware configurations. Balancing the market for most players that are on a budget.

### Steam User Hardware Trends

This section examines the hardware commonly owned by Steam users between 2010 and 2020 using the Steam Hardware Survey dataset. The dataset records the percentage of users with different CPU speeds and RAM capacities

To compare this dataset with the game requirements dataset, CPU speed ranges were converted into numerical midpoint values. Weighted averages were then calculated using the percentage of Steam users represented by each hardware category.

The purpose of this analysis is to determine how consumer hardware evolved during the same period and whether it kept pace with rising game requirements.

```python
# Since the CPU speeds are given as ranges per year we need to convert them to numbers
# To do this we simply added the two numbers and divided by two
import re

def extract_midpoint(cpu_range):
    nums = re.findall(r"\d+\.\d+", str(cpu_range))

    if len(nums) >= 2:
        return (float(nums[0]) + float(nums[1])) / 2

    return None

# Next we find the weighted average CPU speed by year
steam_cpu = Steam_HW_Survey[
    Steam_HW_Survey["category"].isin([
        "AMD CPU Speeds",
        "Intel CPU Speeds"
    ])
].copy()

steam_cpu["cpu_midpoint"] = (
    steam_cpu["name"]
    .apply(extract_midpoint)
)

steam_cpu_yearly = (
    steam_cpu.groupby("year")
    .apply(lambda x:
        (x["cpu_midpoint"] * x["percentage"]).sum()
        / x["percentage"].sum()
    )
)

# We then plot these hardware trends
steam_cpu_yearly.plot(
    figsize=(10,5),
    title="Average Steam User CPU Speed by Year"
)
```

Conversely to the CPU speed requirement of games mid 2010s, the average steam user's speed drops. This decline could have many possibile explanations related to raw performance of hardware or the shift in cultural phenomenons in the gaming industry.

### Hardware Gap Between AAA Games and Steam Users
In this section, we directly compare the minimum hardware requirements of AAA games with the hardware commonly owned by Steam users between 2010 and 2020.

To perform this comparison, yearly median game hardware requirements were merged with yearly weighted average Steam hardware statistics. This allows us to measure the gap between the hardware required to run modern games and the hardware most users actually owned during the same years.

If the gap increases over time, this would support the hypothesis that AAA game requirements grew faster than consumer hardware adoption.

```python
# First we create yearly CPU game requirements 
game_cpu_yearly = (
    processed_df1.groupby("Release_Year")["Min_CPU_CPU_Speed"]
    .median()
    .reset_index(name="median_game_cpu")
)

game_cpu_yearly.head()
```

```python
# We can then merge this with the Steam hardware data
steam_cpu_yearly = (
    steam_cpu.groupby("year")
    .apply(lambda x:
        (x["cpu_midpoint"] * x["percentage"]).sum()
        / x["percentage"].sum()
    )
    .reset_index(name="avg_steam_cpu")
)

comparison = pd.merge(
    game_cpu_yearly,
    steam_cpu_yearly,
    left_on="Release_Year",
    right_on="year"
)

comparison.head()
```

```python
# We can then compute the gap between the median and average of pc users

comparison["cpu_gap"] = (
    comparison["median_game_cpu"]
    - comparison["avg_steam_cpu"]
)

comparison.head()
```

```python
# yearly game requirements
df_games_filtered = processed_df1[(processed_df1['Release_Year'] >= 2010) & (processed_df1['Release_Year'] <= 2020)]

game_ram_yearly = (
    df_games_filtered.groupby("Release_Year")["Min_RAM_MB"]
    .median()
    .reset_index(name="median_game_ram")
)

print(game_ram_yearly.head())

# Plot the Boxplot
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_games_filtered, x='Release_Year', y='Min_RAM_MB', color="#e74c3c")
plt.title("Distribution of Game Minimum RAM Requirements (2010-2020)")
plt.xlabel("Year")
plt.ylabel("Minimum RAM (MB)")
plt.show()
```

In the below code, we aim to show the growth of the amount of RAM a user has over time. First, we extracted numerical values from text strings, as the data was originally labeled as something like "5 GB and Higher". Because the Steam data is provided as percentage shares rather than raw user counts, we calculated a weighted average by multiplying the cleaned RAM sizes by their respective user percentages. Finally, we grouped these weighted values by month, then by year to plot the historical growth of average consumer RAM from 2010 to 2020

```python
# clean the memory
def clean_memory(x):
    if pd.isna(x):
        return np.nan
    
    x = str(x).lower().replace(",", "")
    # find first digits
    match = re.search(r'(\d+\.?\d*)', x)
    
    if not match:
        return np.nan
        
    val = float(match.group(1))

    if "gb" in x:
        return val * 1024
    elif "mb" in x:
        return val
        
    return np.nan

steam_ram = Steam_HW_Survey[Steam_HW_Survey['category'] == 'System RAM'].copy()
steam_ram['RAM_MB'] = steam_ram['name'].apply(clean_memory)
steam_ram['Weighted'] = steam_ram['RAM_MB'] * steam_ram['percentage']

steam_ram_yearly = (
    steam_ram.groupby(['year', 'date'])['Weighted'].sum()
    .groupby('year').mean()
    .reset_index(name="avg_steam_ram")
)

print(steam_ram_yearly.head())

# Plot the average consumer RAM
plt.figure(figsize=(10, 5))
plt.plot(
    steam_ram_yearly["year"], 
    steam_ram_yearly["avg_steam_ram"], 
    marker="o", 
    color="#2ecc71",
    label="Average Steam User RAM"
)
plt.title("Average Consumer System RAM Over Time")
plt.xlabel("Year")
plt.ylabel("System RAM (MB)")
plt.legend()
plt.grid(True)
plt.show()
```

The average amount of RAM a consumer has seems to be growing at a faster rate than the minimum required amount of RAM here. 

```python
comparison_ram = pd.merge(
    game_ram_yearly,
    steam_ram_yearly,
    left_on="Release_Year",
    right_on="year"
)

# Compute the differences between median game RAM and avg steam RAM
comparison_ram["ram_gap"] = (
    comparison_ram["median_game_ram"] 
    - comparison_ram["avg_steam_ram"]
)

# Display the merged dataframe with the new gap column
display(comparison_ram.head())
```

This reveals the differences between the amount of RAM a consumer has, vs the median minimum required amount of RAM for games.

```python
plt.figure(figsize=(12, 6))

# Plot Game RAM
plt.plot(
    comparison_ram["Release_Year"],
    comparison_ram["median_game_ram"],
    marker="o",
    label="Median Game RAM Requirement",
    linewidth=2
)

# Plot Steam RAM
plt.plot(
    comparison_ram["Release_Year"],
    comparison_ram["avg_steam_ram"],
    marker="o",
    label="Average Steam User RAM",
    linewidth=2
)

plt.title("AAA Game RAM Requirements vs Steam User RAM")
plt.xlabel("Year")
plt.ylabel("Memory Size (MB)")
plt.legend()
plt.grid(True)
plt.show()
```

In this visual, we can see that at every point in the 2010s, steam users on average have had more RAM than what is required for the average game. This is about what we had predicted, as the average steam game isn't typically very demanding in terms of RAM, especially when you factor in the number of indie games there are.
